```python
salary = 7000
tithes= 700 #10% of salary
savings = 350 #5% of salary
spendable_amount = 5950
categories = ["food","rent" , "toiletries", "entertainment", "accounts", "transport", "clothing"]
total_spent = 0
```

```python
def get_amount():
  while True:
    text = input("Enter the amount spent on a category: ")
    try:
      amount = float(text)
      if amount > 0:
        return amount
      else:
        print("The amount must be a positive number.")
    except ValueError:
      print("The amount must be a number.")



```

```python
def add_expense(total_spent):
  while True:
    category= input  ("Enter your category (or quit:").strip().lower()
    if category =="quit":
      return total_spent,True
    if category in categories:
        break
    print("That category doesnt exist.Try again")
  amount = get_amount()
  total_spent+=amount

  Remaining = spendable_amount - total_spent
  if Remaining <0:
    print("Your spending has exceeded your monthly spendable amount")
  else:
    print(f" The remaining balance is: R{Remaining:.2f}")
  return total_spent, False



```

```python
total_spent = 0
while True:
  total_spent, finished = add_expense(total_spent)
  if finished:
    break
print(f"Total spent: R{total_spent:.2f}")
```
